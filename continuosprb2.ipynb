## Uniform Distribution

To generate uniform random numbers between any two given values using scipy, we can either use the following code or the code that we have
discussed in class:

```python
from scipy.stats import uniform
```

**Your task:**

1. Based on the code above, write a function that generates uniformly distributed random numbers. There are several requirements for your function:
    * It should accept 3 parameters: 
        * `bottom` - the lower boundary of the generated numbers
        * `ceiling` - the upper boundary of the generated numbers
        * `count` - how many numbers to generate
    * It should return an array of uniformly distributed random numbers

1. Call your function with 2 sets of params below:
    * bottom=10, ceiling=15, count=100
    * bottom=10, ceiling=60, count=1,000

1. Plot the uniform distributions generated above using histograms, where x axis is the value and y axis is the count. Let the histogram's number of bins be 10.

Your output should look like below:

![uniform distribution](ud.png)

```python
def generate_uniform_random_numbers(bottom, ceiling, count):
    return uniform.rvs(loc=bottom, scale=(ceiling - bottom), size=count)
```

```python
set1 = generate_uniform_random_numbers(bottom=10, ceiling=15, count=100)
set2 = generate_uniform_random_numbers(bottom=10, ceiling=60, count=1000)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.hist(set1, bins=10, color='skyblue', edgecolor='black')
plt.title("Uniform Distribution Set 1", fontsize=14)
plt.xlabel("Value", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.ylim(0, 100)  

plt.subplot(1, 2, 2)
plt.hist(set2, bins=10, color='lightgreen', edgecolor='black')
plt.title("Uniform Distribution Set 2", fontsize=14)
plt.xlabel("Value", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.ylim(0, 100) 

plt.tight_layout()
plt.show()
```

How are the two distributions different?

```python
# In the first set we see a small range with few samples which creates sparse with variability in bin heights. 
# In the set 2 we see a wide range with more samples where it provides a smoother and more accurate representation of the distibution. 
```

## Normal Distribution

1. In the same way in the Uniform Distribution challenge, write a function that generates normally distributed random numbers.
1. Generate 1,000 normally distributed numbers with the average of 10 and standard deviation of 1
1. Generate 1,000 normally distributed numbers with the average of 10 and standard deviation of 50
2. Plot the distributions of the data generated.

Expected output:

![normal distribution](nd.png)

```python
from scipy.stats import norm 
```

```python
def generate_normal_random_numbers(mean, std_dev, count):
    return norm.rvs(loc=mean, scale=std_dev, size=count)
```

```python
data1 = generate_normal_random_numbers(mean=10, std_dev=1, count=1000)
```

```python
data2 = generate_normal_random_numbers(mean=10, std_dev=50, count=1000)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.hist(data1, bins=30, color='skyblue', edgecolor='black', density=True)
plt.title("Normal Distribution (mean=10, std=1)", fontsize=14)
plt.xlabel("Value", fontsize=12)
plt.ylabel("Density", fontsize=12)

plt.subplot(1, 2, 2)
plt.hist(data2, bins=30, color='lightgreen', edgecolor='black', density=True)
plt.title("Normal Distribution (mean=10, std=50)", fontsize=14)
plt.xlabel("Value", fontsize=12)
plt.ylabel("Density", fontsize=12)

plt.tight_layout()
plt.show()
```

How are the two distributions different?

## Exponential Distribution

1. Using `numpy.random.exponential`, create a function that returns a list of numbers exponentially distributed with the mean of 10. 

1. Use the function to generate two number sequences with the size of 10 and 100.

1. Plot the distributions as histograms with the nubmer of bins as 100.

Your output should look like below:

![exponential distribution](ed.png)

```python
def generate_exponential_data(mean, size):
    scale = mean  
    return np.random.exponential(scale, size)

sequence_10 = generate_exponential_data(mean=1, size=1000)  
sequence_100 = generate_exponential_data(mean=100, size=1000)

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.hist(sequence_10, bins=100, edgecolor='black', alpha=0.7)
plt.title('Exponential Distribution (Mean=1)', fontsize=14)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xlim(0, 8)  # Adjusting x-axis limit
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.subplot(1, 2, 2)
plt.hist(sequence_100, bins=100, edgecolor='black', alpha=0.7)
plt.title('Exponential Distribution (Mean=100)', fontsize=14)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

How are the two distributions different?

The mean changes, so the distribution changes as well. 

## Exponential Distribution of Real Data

Suppose that the amount of time one spends in a bank is exponentially distributed with mean as 10 minutes (i.e. λ = 1/10). What is the probability that a customer will spend less than fifteen minutes in the bank? 

Write a code in python to solve this problem

```python
mean = 10  # mean time in minutes
lambda_value = 1 / mean 
x = 15  

probability = 1 - math.exp(-lambda_value * x)
```

What is the probability that the customer will spend more than 15 minutes

```python
print(f"The probability that a customer will spend less than 15 minutes in the bank is {probability:.4f}")
```
